```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
from pathlib import Path
import zipfile

import kagglehub
import pandas as pd

DATASET_HANDLE = "datascikhan/e-commerce-sales-and-customer-analytics"

# Download the dataset and locate its first CSV file.
dataset_path = Path(kagglehub.dataset_download(DATASET_HANDLE))

if dataset_path.is_file() and zipfile.is_zipfile(dataset_path):
    with zipfile.ZipFile(dataset_path) as archive:
        csv_names = [name for name in archive.namelist() if name.lower().endswith(".csv")]
        if not csv_names:
            raise FileNotFoundError("No CSV file was found in the downloaded dataset.")
        csv_name = csv_names[0]
        with archive.open(csv_name) as csv_file:
            df = pd.read_csv(csv_file, sep=None, engine="python", encoding="utf-8")
else:
    csv_files = list(dataset_path.rglob("*.csv")) if dataset_path.is_dir() else [dataset_path]
    if not csv_files:
        raise FileNotFoundError("No CSV file was found in the downloaded dataset.")
    df = pd.read_csv(csv_files[0], sep=None, engine="python", encoding="utf-8")

print("Loaded file:", csv_name if "csv_name" in locals() else csv_files[0].name)
print("First 5 records:", df.head())
```
